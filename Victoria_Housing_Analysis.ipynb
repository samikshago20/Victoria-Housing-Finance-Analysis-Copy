```python
import pandas as pd

investor_file = 'Table_14_Investors.xlsx'

investor_excel = pd.ExcelFile(investor_file)

print(investor_excel.sheet_names)
```

```python
investor_meta = pd.read_excel(
    investor_file,
    sheet_name='Index'
)

print(investor_meta.columns.tolist())
```

```python
investor_preview = pd.read_excel(
    investor_file,
    sheet_name='Index',
    header=None
)

print(investor_preview.head(15).to_string())
```

```python
investor_meta = pd.read_excel(
    investor_file,
    sheet_name='Index',
    header=9
)

victoria_investor = investor_meta[
    investor_meta['Data Item Description'].str.contains(
        'Victoria', case=False, na=False
    )
]

print(
    victoria_investor[
        ['Data Item Description', 'Series Type', 'Series ID', 'Unit']
    ].to_string(index=False)
)
```

```python
investor_raw = pd.read_excel(
    investor_file,
    sheet_name='Data1',
    header=None
)

print(investor_raw.head(12).to_string())
```

```python
# Get the Series IDs from row 9
series_ids = investor_raw.iloc[9, 1:]

# Find the column positions for Victoria investor series
count_col = series_ids[series_ids == 'A130268326R'].index[0]
value_col = series_ids[series_ids == 'A130268325L'].index[0]

# Extract dates and values
investor_loans = pd.DataFrame({
    'Date': pd.to_datetime(investor_raw.iloc[10:, 0]),
    'Investor_Loan_Count': investor_raw[count_col].iloc[10:].values,
    'Investor_Loan_Value_Millions': investor_raw[value_col].iloc[10:].values
})

# Keep only 2020–2026
investor_loans = investor_loans[
    (investor_loans['Date'] >= '2020-01-01') &
    (investor_loans['Date'] <= '2026-12-31')
].copy()

print(investor_loans.head())
print(investor_loans.tail())
print(investor_loans.shape)
```

```python
# Convert loan columns to numeric
numeric_columns = [
    'Investor_Loan_Count',
    'Investor_Loan_Value_Millions'
]

investor_loans[numeric_columns] = investor_loans[
    numeric_columns
].apply(pd.to_numeric, errors='coerce')

# Calculate average investor loan size
investor_loans['Average_Investor_Loan_Size'] = (
    investor_loans['Investor_Loan_Value_Millions'] * 1_000_000
    / investor_loans['Investor_Loan_Count']
)

# Calculate quarterly growth
investor_loans['Investor_Count_QoQ_Growth'] = (
    investor_loans['Investor_Loan_Count'].pct_change() * 100
)

investor_loans['Investor_Value_QoQ_Growth'] = (
    investor_loans['Investor_Loan_Value_Millions'].pct_change() * 100
)

# Calculate year-on-year growth (4 quarters)
investor_loans['Investor_Count_YoY_Growth'] = (
    investor_loans['Investor_Loan_Count'].pct_change(periods=4) * 100
)

investor_loans['Investor_Value_YoY_Growth'] = (
    investor_loans['Investor_Loan_Value_Millions'].pct_change(periods=4) * 100
)

# Display the latest 8 quarters
investor_loans.tail(8).round(2)
```

```python
print(investor_loans.columns.tolist())
print(investor_loans.tail(3).to_string(index=False))
print(investor_loans.dtypes)
```

```python
comparison = pd.merge(
    victoria_loans,
    investor_loans,
    on='Date',
    how='inner'
)

print(comparison.shape)
print(comparison.head())
```

```python
# Combined loan commitments
comparison['Total_Loan_Count'] = (
    comparison['Loan_Count']
    + comparison['Investor_Loan_Count']
)

# Combined loan value
comparison['Total_Loan_Value_Millions'] = (
    comparison['Loan_Value_Millions']
    + comparison['Investor_Loan_Value_Millions']
)

# Investor share by number of commitments
comparison['Investor_Count_Share_%'] = (
    comparison['Investor_Loan_Count']
    / comparison['Total_Loan_Count']
) * 100

# Investor share by loan value
comparison['Investor_Value_Share_%'] = (
    comparison['Investor_Loan_Value_Millions']
    / comparison['Total_Loan_Value_Millions']
) * 100

# Display the latest 8 quarters
comparison[
    [
        'Date',
        'Loan_Count',
        'Investor_Loan_Count',
        'Total_Loan_Count',
        'Investor_Count_Share_%',
        'Investor_Value_Share_%'
    ]
].tail(8).round(2)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    comparison['Date'],
    comparison['Investor_Count_Share_%'],
    marker='o',
    label='Investor share by loan count'
)

plt.plot(
    comparison['Date'],
    comparison['Investor_Value_Share_%'],
    marker='o',
    label='Investor share by loan value'
)

plt.title('Investor Share of Housing Loan Commitments – Victoria (2020–2026)')
plt.xlabel('Year')
plt.ylabel('Investor Share (%)')

plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
first_home_file = 'Table_24_First_Home_Buyers.xlsx'

first_home_meta = pd.read_excel(
    first_home_file,
    sheet_name='Index',
    header=9
)

print(first_home_meta.columns.tolist())
print(first_home_meta.head())
```

```python
first_home_raw = pd.read_excel(
    first_home_file,
    sheet_name='Data1',
    header=None
)

# Find the series ID in the Data1 sheet
first_home_series_ids = first_home_raw.iloc[9, 1:]

first_home_count_col = first_home_series_ids[
    first_home_series_ids == 'A130268071X'
].index[0]

# Create a clean DataFrame
victoria_first_home = pd.DataFrame({
    'Date': pd.to_datetime(first_home_raw.iloc[10:, 0]),
    'First_Home_Buyer_Count':
        first_home_raw[first_home_count_col].iloc[10:].values
})

# Keep only 2020–2026
victoria_first_home = victoria_first_home[
    (victoria_first_home['Date'] >= '2020-01-01') &
    (victoria_first_home['Date'] <= '2026-12-31')
].copy()

# Convert to numeric
victoria_first_home['First_Home_Buyer_Count'] = pd.to_numeric(
    victoria_first_home['First_Home_Buyer_Count'],
    errors='coerce'
)

print(victoria_first_home.shape)
print(victoria_first_home.head())
print(victoria_first_home.tail())
```

```python
# Calculate quarterly growth
victoria_first_home['First_Home_Buyer_QoQ_Growth_%'] = (
    victoria_first_home['First_Home_Buyer_Count'].pct_change() * 100
)

# Calculate yearly growth (4 quarters)
victoria_first_home['First_Home_Buyer_YoY_Growth_%'] = (
    victoria_first_home['First_Home_Buyer_Count'].pct_change(periods=4) * 100
)

# Display the latest 8 quarters
print(
    victoria_first_home.tail(8).round(2).to_string(index=False)
)
```

```python
# Merge first-home buyer data with owner-occupier data
comparison = pd.merge(
    comparison,
    victoria_first_home[
        ['Date', 'First_Home_Buyer_Count']
    ],
    on='Date',
    how='inner'
)

# Calculate first-home buyer share
comparison['First_Home_Buyer_Share_%'] = (
    comparison['First_Home_Buyer_Count']
    / comparison['Loan_Count']
) * 100

# Display latest 8 quarters
print(
    comparison[
        [
            'Date',
            'Loan_Count',
            'First_Home_Buyer_Count',
            'First_Home_Buyer_Share_%'
        ]
    ].tail(8).round(2).to_string(index=False)
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    comparison['Date'],
    comparison['First_Home_Buyer_Share_%'],
    marker='o'
)

plt.title('First Home Buyer Share of Owner-Occupier Loans – Victoria')
plt.xlabel('Year')
plt.ylabel('First Home Buyer Share (%)')

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
comparison = pd.merge(
    comparison,
    victoria_first_home[
        [
            'Date',
            'First_Home_Buyer_QoQ_Growth_%',
            'First_Home_Buyer_YoY_Growth_%'
        ]
    ],
    on='Date',
    how='left'
)

print(comparison.shape)
print(comparison.columns.tolist())
```

```python
print("Dataset shape:", comparison.shape)

print("Duplicate dates:", comparison['Date'].duplicated().sum())

print("Missing values:")
print(comparison.isnull().sum())

print("Data types:")
print(comparison.dtypes)
```
